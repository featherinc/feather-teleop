# Quest teleoperation

Run on the robot's computer; the controls are in the README. Worn: once the first cell has run, open `https://<robot address>:8443` in the headset's browser. Table with `source: network`: keep `python -m feather_teleop.headsets.table --to <robot address>` running on the laptop the headset is plugged into.

```python
from feathersdk.robot.robot import initialize_feather_system

from feather_teleop import config
from feather_teleop.teleop import QuestTeleop

settings = config.load()
robot = initialize_feather_system(config.sdk_config(settings))
teleop = QuestTeleop(robot, settings)
```

**Moves the robot.** The torso rises, then the arms go to the start posture. This move is not collision checked: clear the space around the arms.

```python
teleop.ready()
```

**Moves the robot.** Press **A** to start a demonstration, **A** again to end it as a success, **B** as a failure. The stop button ends the cell and saves an open demonstration as interrupted.

```python
teleop.run()
```

**Moves the robot.** Home the arms and shut down. Run this cell even if the one above ended with an error.

```python
try:
    teleop.home()
finally:
    teleop.close()
    robot.abort()   # the SDK's shutdown: it also runs when the kernel exits
```
